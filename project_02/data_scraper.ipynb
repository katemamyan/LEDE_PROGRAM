## Data scraping part

```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
import pandas as pd

#setup
pd.set_option('display.max_colwidth', 100)
```

```python
url = "https://www.primeminister.am/en/foreign-visits"
r = requests.get(url)
```

```python
# saving html of the main page here
doc = BeautifulSoup(r.text)
```

```python
# finding the last page on a pagination to create links to all pages' links

last_page = int(doc.find_all("li", class_="paging__item ver-top-box")[-1].text.replace("\n", ""))
last_page
```

```python
# Creating a list of all pages

pages = []
for i in range(last_page):
    pages.append(f'https://www.primeminister.am/en/foreign-visits/page/{i+1}/')
pages
```

```python
# creating a function to scrape the visit_links from the given page

def page_scraper(page_link):
    r = requests.get(page_link)
    doc = BeautifulSoup(r.text)
    list_of_a = doc.find_all("a", class_="search__result-title db")
    visit_links = [f'https://www.primeminister.am{a.get("href")}' for a in list_of_a]
    return visit_links
```

```python
# testing with long texts

page_scraper
```

```python
# running the fuction on all pages to get direct link to all pages about visits

all_visit_links = []
for page in pages:
    all_visit_links.extend(page_scraper(page))

all_visit_links
```

```python
len(all_visit_links)
```

```python
# creating a function to scrape the visit_links from the given page


def data_parser(link):
    visit_data = {}
    r = requests.get(link)
    doc = BeautifulSoup(r.text)
    visit_data['link']=link
    visit_data['title']=doc.find("h3").text
    visit_data['date']=doc.find("div", class_="search__date fs12").text
    try:
        visit_data['text']=doc.find("div", class_="news-text").text.replace("\n", "")
    except:
        visit_data['text']="no text in the"
    return visit_data

    
    
```

```python
data_parser("https://www.primeminister.am/en/foreign-visits/item/2020/02/13/Nikol-Pashinyan-visit-to-Germany/")
```

```python
data = []
for link in all_visit_links:
    data.append(data_parser(link))
df = pd.DataFrame(data)
df
```

```python
df.to_csv("scraped_data_en.csv", index=False)
```

```python
import pandas as pd
from openai import OpenAI
import tiktoken  # for counting tokens

# Tokenizer for gpt-3.5-turbo
encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")

MAX_TOKENS = 16385
SAFETY_MARGIN = 500  # leave room for system prompt, etc.
MAX_INPUT_TOKENS = MAX_TOKENS - SAFETY_MARGIN

def truncate_text(text, max_tokens=MAX_INPUT_TOKENS):
    """Truncate text to fit within allowed token limit and return flag."""
    tokens = encoding.encode(text)
    truncated = False
    if len(tokens) > max_tokens:
        tokens = tokens[:max_tokens]
        truncated = True
    return encoding.decode(tokens), truncated

def classify(text):
    try:
        short_text, truncated = truncate_text(text)
        response = client.chat.completions.create(
            model="gpt-4o",
            messages=[
                {
                    "role": "system",
                    "content": (
                        "Please read the text abot Armenina PM Nikol Pashinyan's officials vistis abroad."
                        "Identify the country and city/cities that he visit.\n"
                        "Return your answer in this exact format:\n"
                        "Country name | City1;City2;City3\n"
                        "- Put the international country name, without abbreviations\n"
                        "- Country of visit can not be Armenia!! Please do not write Armenia as a country"
                        "- If country unknown, write: manual check\n"
                        "- If city unknown, write: manual check\n"
                        "- Do not add any extra words to the answer."
                    )
                },
                {"role": "user", "content": short_text}
            ],
            temperature=0,
            top_p=1,
            max_tokens=30
        )
        return response.choices[0].message.content.strip(), truncated
    except Exception as e:
        print(f"Error with message: {text[:50]}... — {e}")
        return "error | error", False

# Load dataset
df = pd.read_csv("/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_02/scraped_data_en.csv")

# Apply classifier & store raw result + truncation flag
df[["raw_result", "was_truncated"]] = df["text"].apply(
    lambda msg: pd.Series(classify(str(msg)))
)

# Split raw_result into country and city columns
split_cols = df["raw_result"].str.split("|", n=1, expand=True)
df["country"] = split_cols[0].str.strip().str.lower()
df["city"] = split_cols[1].str.strip().str.lower() if split_cols.shape[1] > 1 else None

# Save results
output_path = "/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_02/classified_data.csv"
df.to_csv(output_path, index=False)

print(f"Classification completed. Results saved to: {output_path}")
```

## Data cleaning part

```python
# opening 

df = pd.read_csv("/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_02/new_classified_data.csv", index_col=False)
df.head()
```

```python
#creating start and end dates

df[['start_date', 'end_date']] = df['date'].str.split(' - ', expand=True)
df.head()
```

```python
# turning into datetime format
 
df['start_date'] = pd.to_datetime(df['start_date'], format='%d.%m.%Y')
df['end_date'] = pd.to_datetime(df['end_date'], format='%d.%m.%Y')
```

```python
# splitting the City_country column into two separate columns and stripping to remove spaces

# Safely split into two columns, fill missing values
split_cols = df['city_country'].str.split(';', n=1, expand=True)

df['city'] = split_cols[0].str.strip()
df['country'] = split_cols[1].str.strip() if split_cols.shape[1] > 1 else ""
```

```python
df.head(3)
```

```python
# calculating the difference of dates

df['days_diff'] = (pd.to_datetime(df['end_date']) - pd.to_datetime(df['start_date'])).dt.days
df.head(3)
```

```python
# adding year to data

df['year'] = pd.to_datetime(df['start_date']).dt.year
df
```

```python
pd.pivot_table(
    df,
    values='link',            # the column to "count"
    index='country',          # rows
    aggfunc='count',           # counts non-null values (like COUNTA)
).sort_values(by='link', ascending=False)
```

```python
# cleaning dataframe 

df['country'] = df['country'].replace({'russian federation': 'russia', 
                                       'türkiye':'turkey',
                                       'united states of america':'united states',
                                       'holy see':'vatican',
                                       "people's republic of china":'china',
                                       })
```

```python
pd.pivot_table(
    df,
    values='link',            # the column to "count"
    index='country',          # rows
    aggfunc='count',           # counts non-null values (like COUNTA)
).sort_values(by='link', ascending=False)
```

```python

# Pivot table: count 'link' by country and year
pivot = pd.pivot_table(
    df,
    values='link',             # the column to count
    index='country',           # rows
    columns='year',            # years as columns
    aggfunc='count',           # counts non-null values
    fill_value=0               # replace NaN with 0
).sort_values(by=df['year'].max(), ascending=False)  # sort by latest year column
```
