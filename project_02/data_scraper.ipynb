## Data scraping part

```python
import requests
import pandas as pd 
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://www.primeminister.am/en/foreign-visits"
r = requests.get(url)
```

```python
# saving html of the main page here
doc = BeautifulSoup(r.text)
```

```python
# finding the last page on a pagination to create links to all pages' links

last_page = int(doc.find_all("li", class_="paging__item ver-top-box")[-1].text.replace("\n", ""))
last_page
```

```python
# Creating a list of all pages

pages = []
for i in range(last_page):
    pages.append(f'https://www.primeminister.am/en/foreign-visits/page/{i+1}/')
pages
```

```python
# creating a function to scrape the visit_links from the given page

def page_scraper(page_link):
    r = requests.get(page_link)
    doc = BeautifulSoup(r.text)
    list_of_a = doc.find_all("a", class_="search__result-title db")
    visit_links = [f'https://www.primeminister.am{a.get("href")}' for a in list_of_a]
    return visit_links
```

```python
# running the fuction on all pages to get direct link to all pages about visits

all_visit_links = []
for page in pages:
    all_visit_links.extend(page_scraper(page))

all_visit_links
```

```python
len(all_visit_links)
```

```python
# creating a function to scrape the visit_links from the given page


def data_parser(link):
    visit_data = {}
    r = requests.get(link)
    doc = BeautifulSoup(r.text)
    visit_data['link']=doc.find("h3").text
    visit_data['date']=doc.find("div", class_="search__date fs12").text
    try:
        visit_data['text']=doc.find("div", class_="news-text").text.replace("\n", "")
    except:
        visit_data['text']="no text in the"
    return visit_data

    
    
```

```python
data = []
for link in all_visit_links:
    data.append(data_parser(link))
df = pd.DataFrame(data)
df
```

```python
df.to_csv("scraped_data_en.csv")
```

```python
# summaries for text column


# Initialize OpenAI client
# Load new dataset
# df = pd.read_csv("/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_02/scraped_data_en.csv")

# Function to classify drug ads
def classify(text):
    try:
        response = client.chat.completions.create(
            model="gpt-3.5-turbo",
            messages=[
                {
                    "role": "system",
                    "content": (
                        "Provide 2 sentence summary about the visit."
                    )
                },
                {
                    "role": "user",
                    "content": text
                }
            ],
            temperature=0.0,
            max_tokens=20,
            top_p=1
        )
        return response.choices[0].text.content.strip().lower()
    except Exception as e:
        print(f"Error with message: {text[:50]}... — {e}")
        return "error"

# Apply the classifier to each message
df["summary"] = df["text"].apply(lambda msg: classify(str(msg)))

# Save the results to a new CSV
output_path = "/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_02/data_with_summaries"
df.to_csv(output_path, index=False)
```
