```python
import pandas as pd
```

```python
# openning file with data
df = pd.read_csv('/Users/katemamyan/Documents/GitHub/LEDE_PROGRAM/project_01/archive_data.csv', index_col=False)
df.head()
```

```python
# cleaning the date column, removing armenian abbreviations and converting to the date format


# armenian month abbreviations to month numbers
armenian_months = {
    'Հնվ': '01', 'Փտր': '02', 'Մար': '03', 'Ապր': '04', 'Մայ': '05', 'Հնս': '06',
    'Հլս': '07', 'Օգս': '08', 'Սեպ': '09', 'Հոկ': '10', 'Նմբ': '11', 'Դեկ': '12'
}

# Split and map months
df['day'] = df['date'].str.extract(r'(\d{2})')
df['arm_month'] = df['date'].str.extract(r'(\w{3})')
df['year'] = df['date'].str.extract(r'(\d{4})')

df['month'] = df['arm_month'].map(armenian_months)

# Combine into ISO format date
df['iso_date'] = pd.to_datetime(df['year'] + '-' + df['month'] + '-' + df['day'])

print(df[['date', 'iso_date']])
```

```python
df
```

```python
# creating new 'cleaned_title' column where we are cleaning names to keep only the name of the guest

df['cleaned_title'] = (
    df['title']
    .str.replace(
        r'Հարցազրույց Պետրոս Ղազարյանի հետ.|Հարցազրույց|հետ|Պետրոս Ղազարյանի հարցազրույցը|ՀՀ Ազգային ժողովի նախագահ|ՀՀ վարչապետ|փոխվարչապետ',
        '',
        regex=True
    )
    .str.replace(r'ի\b', '', regex=True)
    .str.lower()
    .str.strip()
)
```

```python
df.to_csv("cleaned_archive_data.csv", index=False)
```
