### Better way: CURL

Documentation here: https://github.com/curl/curl

Format:
curl http://data.rcc-acis.org/StnData\?sid\={STATION_ID}\&sdate\={START_DATE}\&edate\={END_DATE}\&elems\={ELEMENT}\&output\=json -o {DIRECTORY}/{FILE-NAME}.json

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline
```

### Example: Station data for NYC (Central Park), from Jan. 1, 1970, through Dec. 31, 2024
curl http://data.rcc-acis.org/StnData\?sid\=USW00094728\&sdate\=1970-01-01\&edate\=2024-12-31\&elems\=maxt\&output\=json -o/Users/veronicapenney/Desktop/lede-2025-data/07-22-working-with-sensor-data/temperature_data/nyc_max_daily_temp.json

```python
import json # this will let us work with our json file
json_data = 'temperature_data/nyc_max_daily_temp.json' # specify the file path for our data
```

```python
with open(json_data, 'r') as f:
    data = json.load(f)
```

```python
# print the file to see what we're working with
data
```

### The metadata in this file describes the station, but doesn't tell us anything about our data
That's okay, because we know what we asked for in our curl request.
Let's ignore the metadata and just pull in the data.

```python
# let's access our data 
data['data'] # this gives us a list of lists
```

```python
# we're working in Pandas, so we want our data in a dataframe
df = pd.DataFrame(data['data'])
df.head()
```

```python
# Add in column names
df.columns = ['date', 'maxt']
df
```

```python
# Check the data types
df.dtypes
```

```python
# convert dates to datetime
df.date = pd.to_datetime(df.date)
# convert average temp to numeric
df.maxt = pd.to_numeric(df.maxt)
df.dtypes
```

```python
print(df.avgt.isnull().sum()) # no NaNs! great
print(len(df))
```

```python
# print(df.describe())
print(df.avgt.describe())
```

```python
df.plot(x='date', y='maxt', kind='scatter')
```

```python
# calculate the annual rate of change
# first, we need years
df['year'] = df['date'].dt.year

import numpy as np
# now, we can calculate the annual rate of change
np.polyfit(df.year, df.avgt, 1)

# # Calculate the average temperature for each year
# yearly_avg_temp = df.groupby('year')['avgt'].mean().reset_index()

# # Convert the year and temperature to NumPy arrays for linear regression
# years = yearly_avg_temp['year'].to_numpy()
# avg_temps = yearly_avg_temp['avgt'].to_numpy()

# yearly_avg_temp.head(3)
```

```python
# we can name the coefficients as variables and then call them
slope, intercept = np.polyfit(df.year, df.avgt, 1)
```

```python
# here is our annual rate of change
print(slope, intercept)
```

```python
# and here is the change in average annual temperature between 1970 and 2024
slope*(2024-1970)
```

```python
slope*(df.year.max() - df.year.min())
```
