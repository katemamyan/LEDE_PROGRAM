# Analyzing SO2 data in 2024
SO2, or sulfur dioxide, is emitted from burning fossil fuels. It's most commonly emitted by power plants, ships, trucks, and volcanoes.

#### There are three federal primary standards for sulfur dioxide (SO2) - 
#### a 1-hour standard of <b>75 parts per billion (ppb)</b>, 
#### a 24-hour standard of <b>0.14 parts per million (ppm)</b>, 
#### and an <b>annual average standard of 0.030 ppm.</b>

```python
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
pd.options.display.max_columns = 50 # we know this is going to be a big dataset

daily_so2 = pd.read_csv('EPA_data/daily_so2_2024.csv') # read in data on sulfur dioxide
hourly_so2 = pd.read_csv('EPA_data/hourly_42401_2024.csv') # read in the hourly measurements for sulfur dioxide
```

```python
hourly_so2.head(3)
```

```python
# are we measuring the correct standard? check
# hourly data should be in parts per billion
hourly_so2['Units of Measure'].unique()
```

```python
hourly_so2['Sample Measurement'].sort_values(ascending=False) # why are we seeing negative values?
# https://www.epa.gov/sites/default/files/2017-02/documents/negative_values_reporting_to_aqs_10_6_16.pdf
```

```python
# do we need to account for uncertainty in the data?
hourly_so2.Uncertainty.unique()
```

```python
# groupby objects are helpful here. we can use all the identifying features to group measurements by their station
hourly_so2.groupby(['State Code', 'County Code', 'Site Num', 'Latitude', 'Longitude'])['Sample Measurement'].head(3)
```

```python
# let's look at hourly measurements by sensor
# there are five identifying features for each station, so let's set those as a variable that we can reuse
identifying_features = ['State Code', 'County Code', 'Site Num', 'Latitude', 'Longitude']
```

```python
# First, filter out the negative values
hourly_so2 = hourly_so2[hourly_so2['Sample Measurement'] >= 0]

# Then, groupby the unique identifiers and get a sense of the reported measurement data
hourly_so2[hourly_so2['Sample Measurement'] >= 0].groupby(identifying_features)['Sample Measurement'].agg(['min','max', 'mean'])
```

```python
# Find every hourly measurement where the air quality measurement exceeded EPA standards
hourly_so2[hourly_so2['Sample Measurement'] > 75]
# Count the number of measurements in excedence of air quality guidelines for each monitor
hourly_so2[hourly_so2['Sample Measurement'] > 75].groupby(identifying_features)['Sample Measurement'].max().reset_index()
```

```python
# export the number of violations per monitor to a csv for mapping
hourly_so2[hourly_so2['Sample Measurement'] > 75].groupby(identifying_features)['Sample Measurement'].count().reset_index().to_csv('EPA_data/processed/so2_hourly_violation_count.csv')
```

```python
# may also be interesting to explore the max exceedence
hourly_so2[hourly_so2['Sample Measurement'] > 75].groupby(identifying_features)['Sample Measurement'].max().reset_index()
# export this to a csv as well
hourly_so2[hourly_so2['Sample Measurement'] > 75].groupby(identifying_features)['Sample Measurement'].max().reset_index().to_csv('EPA_data/processed/so2_max_exceedence_by_sensor.csv')
```

It looks like all the stations are reporting both 1-hour and 3-hour block averages.<br>
The 1-hour measurements are EPA standard, so we'll use those.

#### Constructing the API call
Things we need:<br>
A parameter. We're looking at ozone, which is 44201.<br>
A state number. It's 08<br>
A county number. That's 031<br>
A site number. It's 0026<br>
Dates. We'll start by testing all of 2024.

```python
# Great! Let's get the past five years of data
url_2020_2024 = 'https://aqs.epa.gov/data/api/annualData/byCounty?email=test@aqs.api&key=test&param=44201&bdate=2020101&edate=20241231&state=08&county=031'
r = requests.get(url_2020_2024)

r.json()
```

```python
import requests
import time

start_year = 2020
end_year = 2024

joined = []
for year in range(start_year, end_year + 1):
    url = f'https://aqs.epa.gov/data/api/annualData/byCounty?email=test@aqs.api&key=test&param=44201&bdate={start_year}0101&edate={end_year}1231&state=08&county=031'
    
    print(f'Grabbing {year} file: {url}')
    r = requests.get(url)
    
    api_json = r.json()

    # the Header is an array of lists, so you have to access the first list item to get to the 'status' value
    if api_json['Header'][0]['status'] != 'Success': 
        print('uh-oh')
    else:
        joined += api_json['Data']

    print('Waiting one second to be nice to the EPA servers')
    time.sleep(1)

len(joined)
```
