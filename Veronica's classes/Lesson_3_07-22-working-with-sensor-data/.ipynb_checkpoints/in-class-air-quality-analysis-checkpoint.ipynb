### We'll analyze SO2 data in 2024
SO2, or sulfur dioxide, is emitted from burning fossil fuels. It's most commonly emitted by power plants, ships, trucks, and volcanoes.

#### There are three federal primary standards for sulfur dioxide (SO2) - 
#### a 1-hour standard of <b>75 parts per billion (ppb)</b>, 
#### a 24-hour standard of <b>0.14 parts per million (ppm)</b>, 
#### and an <b>annual average standard of 0.030 ppm.</b>
